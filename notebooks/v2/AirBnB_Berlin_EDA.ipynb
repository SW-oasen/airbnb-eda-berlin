### The Occupancy Model

One of the biggest issues with Airbnb is whether hosts are renting out residential properties permanently as hotels, rather than sharing their primary residence "occasionally." An occupancy model estimates how often a listing is rented and approximates its income.

Inside Airbnb uses an occupancy model named the "**San Francisco Model**", honoring public policy and urban planners who quantified Airbnb's impact on housing.

### Key Elements of the Model

1. ** Review Rate:**
   - A Review Rate of 50% is used to convert reviews into estimated bookings, assuming only 50% of guests leave reviews.
   - Other review rates considered:
     - **72%**: Attributed to Airbnb's CEO but unverifiable.
     - **30.5%**: Based on comparisons with public data and the New York Attorney General’s report (2014).
   - Inside Airbnb chose 50% as a conservative midpoint between 72% and 30.5%.

2. ** Average Length of Stay:**
   - Multiplied by estimated bookings to determine the occupancy rate:
     - **San Francisco**: 5.5 nights per stay (reported by Airbnb).
     - **Other Cities**: 3 nights per booking (default if data is unavailable).
     - Higher minimum nights values override these averages.

3. ** Occupancy Rate Cap:**
   - Capped at **70%** to maintain conservatism and reflect a reasonable maximum for highly occupied "hotels."
   - Adjusts for:
     - Seasonal changes affecting minimum nights.
     - Listings with very high review rates.

4. ** Filtering Metrics:**
   - Aligns the number of nights booked or available per year with city short-term rental laws to protect residential housing.

By using this model, Inside Airbnb provides insights into listing activity and its potential impact on housing and neighborhoods.

## Filtering Listings for Analysis 

In this section, the dataset is filtered to focus on listings that are more likely to be actively booked and relevant to the analysis. 
- Focus on listings with **reviews per month > 2** (indicative of demand).
- Include only listings with **minimum nights <= 5** (removing highly restrictive options).
- Filter prices to **< $500** to exclude outliers.

```python
# Importing the necessary library
import pandas as pd
from IPython.display import display

# Set pandas display options to show all rows and columns
pd.set_option('display.max_rows', None)  # Show all rows
pd.set_option('display.max_columns', None)  # Show all columns

print('Dataset before filtering:\n', df.shape)

# Clean Price: Convert to Integer - Removes symbols $ and %, handles decimals, and converts to int
df['price'] = (
    df['price']
    .replace(r'[\$,€%]', '', regex=True)
    .replace(',', '', regex=True)
    .astype(float)
    .fillna(0)
    .astype(int)
    )

def filter_active_market(df):
    """
    Filters the dataset for high-demand, short-term, and moderately priced listings.
    Compares the size of the dataset before and after.
    """
    # 1. Capture initial size
    initial_count = len(df)
    
    # 2. Apply Filters
    # - reviews_per_month > 2: High frequency/Active demand
    # - minimum_nights <= 5: True short-term rental behavior
    # - price < 500: Typical market range (excluding luxury/outliers)
    
    # Using the numeric 'price' column created in previous steps
    df_filtered = df[
        (df['reviews_per_month'] > 2) & 
        (df['minimum_nights'] <= 5) & 
        (df['price'] < 500)
    ].copy()
    
    # 3. Capture final size
    final_count = len(df_filtered)
    reduction = initial_count - final_count
    reduction_pct = (reduction / initial_count) * 100

    # 4. Print Comparison
    print("--- Dataset Filtering Summary ---")
    print(f"Initial Listings:      {initial_count}")
    print(f"Filtered (Active):     {final_count}")
    print(f"Listings Removed:      {reduction} ({reduction_pct:.2f}%)")
    print("-" * 33)
    
    return df_filtered

# --- EXECUTION ---
# Note: Use the revenue_df created previously which has the numeric 'price' column.
df_filtered = filter_active_market(df)

print('Dataset after filtering:\n', df_filtered.shape)

#display(df_filtered.head())
```

## Feature selection - Filter relevant columns for analysis

Carrying around 75+ columns of raw text (descriptions, house rules, URLs) is a waste of RAM and makes debugging much harder. It creates "cognitive noise." For the revenue analysis only key metrics are selected. 

```python
import pandas as pd

def slim_down_dataset(df):
    """
    Retains only the columns necessary for revenue, geographic, and room-type analysis.
    Drops URLs, long text descriptions, and scraping metadata.
    """
    # Define the 'Golden Columns' we actually need
    keep_columns = [
        'id', 
        'name', 
        'host_id',
        'host_name',
        'neighbourhood_cleansed', 
        'neighbourhood_group_cleansed', 
        'latitude', 
        'longitude',
        'property_type', 
        'room_type', 
        'price',               # The cleaned numeric price
        'minimum_nights', 
        'number_of_reviews', 
        'reviews_per_month',
        'availability_365',
        'license',             # Important for Berlin regulatory analysis
        'revenue_proxy_lower', # Our calculated metric
        'revenue_proxy_upper'  # Our calculated metric
    ]
    
    # Use intersection to avoid errors if a column was already dropped or renamed
    existing_cols = [c for c in keep_columns if c in df.columns]
    
    df_slim = df[existing_cols].copy()
    
    # Print impact report
    print("--- Memory Optimization Summary ---")
    print(f"Columns before: {df.shape[1]}")
    print(f"Columns after:  {df_slim.shape[1]}")
    
    # Calculate memory reduction
    mem_before = df.memory_usage(deep=True).sum() / 1024**2
    mem_after = df_slim.memory_usage(deep=True).sum() / 1024**2
    print(f"Memory Usage:   {mem_before:.2f} MB -> {mem_after:.2f} MB")
    print(f"Reduction:      {((mem_before - mem_after) / mem_before)*100:.1f}%")
    print("-" * 35)
    
    return df_slim

# --- EXECUTION ---
# Run this on your active market dataframe
df_filtered = slim_down_dataset(df_filtered)

# Now when you look at the head, it's clean and analytical
print(df_filtered.head())
```

## Cleaning the data

However the column 'license' still contains wrong data. The categorical columns need to be changed to correct dtypes. 

```python
import pandas as pd
import re

def final_data_cleaning(df):
    """
    1. Sanitizes the 'license' column into meaningful categories.
    2. Converts string columns to 'category' dtypes for performance.
    """
    df_clean = df.copy()

    # --- 1. LICENSE SANITIZATION ---
    # Berlin Registration Format: 01/Z/ZA/001234-15 (or similar patterns with /Z/)
    # We categorize them as: 'Valid Format', 'Exempt', 'Illegal/Dirty'
    
    def classify_license(val):
        val = str(val).lower().strip()
        if val in ['nan', 'none', '', 'null']:
            return 'Missing'
        # Regular expression for the official Berlin format:
        # Looking for the signature '/Z/' or '/Z/ZA/'
        if re.search(r'\d{2}/z/', val):
            return 'Valid License'
        # Check for exemptions
        if any(keyword in val for keyword in ['exempt', 'befreiung', 'gewerbe', 'hotel', 'pension']):
            return 'Exempt'
        # Everything else (like "private", "12345", "no license")
        return 'Invalid/Dirty'

    df_clean['license_status'] = df_clean['license'].apply(classify_license)

    # --- 2. DTYPE CONVERSION (Optimizing for Categories) ---
    cat_cols = [
        'neighbourhood_cleansed', 
        'neighbourhood_group_cleansed', 
        'property_type', 
        'room_type', 
        'license_status'
    ]
    
    for col in cat_cols:
        if col in df_clean.columns:
            df_clean[col] = df_clean[col].astype('category')

    # Ensure numeric columns are strictly floats/ints
    numeric_cols = ['price', 'minimum_nights', 'reviews_per_month', 'availability_365']
    for col in numeric_cols:
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

    # --- 3. FINAL SUMMARY ---
    print("--- Final Cleaning & Dtype Summary ---")
    print(df_clean.info())
    print("\nLicense Status Distribution:")
    print(df_clean['license_status'].value_counts())
    
    return df_clean

# --- EXECUTION ---
df_clean = final_data_cleaning(df_filtered)

# df_final_clean.to_csv('airbnb_berlin_final_cleaned_20250923.csv', index=False)
# csv export for Power BI with specific formatting with German locale 
#df_clean.to_csv('berlin_airbnb_20250923_powerbi_de.csv', sep=';', decimal=',', index=False, encoding='utf-8-sig')

# explore the final cleaned data
explore_airbnb_data(df_clean)
```

## Revenue Analysis 

In this section, we aim to address the business question:

**Which property types and neighborhoods generate the highest revenue?**

To answer this, we will use a **revenue proxy** defined as:



### Steps in the Calculation

1. **Estimate Bookings**:
   Scale the number of reviews by the `review_rate` (e.g., 30% or 70%) to estimate the total number of bookings:

   

2. **Calculate Maximum Bookings**:
   Compute the maximum allowable bookings based on a **90% occupancy cap** and the ratio of average days per month (`days_per_month`) to `minimum_nights`:

    

3. **Normalize Bookings**:
   Take the smaller of the two values (i.e., `estimated_bookings` and `max_bookings`) to account for realistic occupancy limits:
   
    

4. **Calculate Revenue Proxy**:
   Multiply the `normalized_bookings` by the `price` to compute the final revenue proxy:
   
   

### Application of Revenue Proxy

We will calculate two columns using different `review_rate` values:
- **`revenue_proxy_upper`**: Using a review rate of 30%.
- **`revenue_proxy_lower`**: Using a review rate of 70%.

### Inspection Table

To inspect the data and identify any outliers or property types that may not align well with the model, we will generate a table with the following columns:
- `Listing name`
- `property_type`
- `neighbourhood_cleansed`
- `minimum_nights`
- `reviews_per_month`
- `total_reviews`
- `price`

This breakdown will help us understand how revenue varies across property types and neighborhoods.

```python
import pandas as pd
import numpy as np

def calculate_berlin_revenue(df):
    # 1. Clean Price: Convert to Integer - done before

    # 2. Setup Constants for the Model 
    DAYS_PER_MONTH = 30.44
    OCCUPANCY_CAP = 0.90  # 90% as per step 2
    
    # Thresholds for Review Rate
    # Lower review rate = Higher estimated revenue (more un-reviewed stays)
    REVIEW_RATE_UPPER_BOUND = 0.30 
    REVIEW_RATE_LOWER_BOUND = 0.70

    # 3. Intermediate Calculations
    # Calculate Maximum Bookings based on 90% occupancy cap and minimum nights
    # If min_nights is 30, max_bookings will be approx 0.9 per month
    df['max_bookings'] = OCCUPANCY_CAP * (DAYS_PER_MONTH / df['minimum_nights'])

    # 4. Revenue Proxy Upper (30% review rate)
    df['est_bookings_upper'] = df['reviews_per_month'].fillna(0) / REVIEW_RATE_UPPER_BOUND
    df['norm_bookings_upper'] = df[['est_bookings_upper', 'max_bookings']].min(axis=1)
    df['revenue_proxy_upper'] = (df['price'] * df['norm_bookings_upper']).round(2)

    # 5. Revenue Proxy Lower (70% review rate)
    df['est_bookings_lower'] = df['reviews_per_month'].fillna(0) / REVIEW_RATE_LOWER_BOUND
    df['norm_bookings_lower'] = df[['est_bookings_lower', 'max_bookings']].min(axis=1)
    df['revenue_proxy_lower'] = (df['price'] * df['norm_bookings_lower']).round(2)

    # Sort by the upper proxy to find high-earners/outliers
    return df.sort_values(by='revenue_proxy_upper', ascending=False)

df_final = calculate_berlin_revenue(df_clean)
print(df_final.head(20))

# Export final cleaned data for Power BI with specific formatting of German locale
df_final.to_csv('berlin_airbnb_20250923_revenue_powerbi_de.csv', sep=';', decimal=',', index=False, encoding='utf-8-sig')
```

## Check Outliers

Outliers are data points that deviate significantly from the rest of the dataset, potentially influencing the results of analysis. Identifying and understanding outliers can help improve the robustness of your insights.

Criteria:
- Revenue Proxy Upper > €15,000/month
- OR Nightly Price > €1,500, which are already filtered out in a previous step

```python
# Outliers can be flagged with a threshold (e.g., excessively high proxies) or highly revenue generating listings!!

def identify_revenue_outliers(df, revenue_threshold=15000, price_threshold=1500):
    """
    Filters the revenue table for implausible listings.
    Criteria:
    - Revenue Proxy Upper > €15,000/month
    - OR Nightly Price > €1,500
    """
    
    # Identify outliers based on revenue and price
    outliers = df[
        (df['revenue_proxy_upper'] > revenue_threshold) | 
        (df['price'] > price_threshold)
    ].copy()

    # Add a 'Reason' column for analysis
    conditions = [
        (outliers['price'] > price_threshold),
        (outliers['revenue_proxy_upper'] > revenue_threshold) & (outliers['number_of_reviews'] < 5),
        (outliers['revenue_proxy_upper'] > revenue_threshold)
    ]
    choices = ['Extreme Nightly Price', 'High Revenue / Low Data (New Listing)', 'High Revenue Model Outlier']
    outliers['outlier_reason'] = np.select(conditions, choices, default='Other')

    # Sort by the most extreme revenue estimates
    return outliers.sort_values(by='revenue_proxy_upper', ascending=False)

outlier_table = identify_revenue_outliers(df_final)

# To view the columns requested:
print(outlier_table[[
    'name', 'property_type', 'neighbourhood_cleansed', 
    'minimum_nights', 'reviews_per_month', 'number_of_reviews', 
    'price', 'revenue_proxy_lower', 'revenue_proxy_upper', 'outlier_reason'
 ]].head(15))
```

```python
import pandas as pd

def generate_revenue_summary(df, proxy_col):
    """
    Generates a summary table for a specific revenue proxy column.
    """
    # Grouping and Aggregating
    summary = df.groupby(['neighbourhood_cleansed', 'room_type']).agg(
        total_listings=('name', 'count'),
        average=(proxy_col, 'mean'),
        lower_quantile=(proxy_col, lambda x: x.quantile(0.25)),
        upper_quantile=(proxy_col, lambda x: x.quantile(0.75)),
        avg_price=('price', 'mean'),
        avg_minimum_nights=('minimum_nights', 'mean')
    ).reset_index()

    # Rounding for readability
    summary = summary.round(2)

    # Sorting by average revenue and then total listings (descending)
    summary_sorted = summary.sort_values(
        by=['average', 'total_listings'], 
        ascending=[False, False]
    )

    return summary_sorted

# --- STEP 1: Generate Tables ---
summary_upper = generate_revenue_summary(df_final, 'revenue_proxy_upper')
summary_lower = generate_revenue_summary(df_final, 'revenue_proxy_lower');

# --- STEP 2: Display Top 20 for Upper Proxy ---
print("### TOP 20 NEIGHBORHOOD/ROOM TYPE BY REVENUE PROXY (UPPER) ###")
#print(summary_upper.head(20).to_string(index=False))
print(summary_upper.head(20))

print("\n" + "="*80 + "\n")

# --- STEP 3: Display Top 20 for Lower Proxy ---
print("### TOP 20 NEIGHBORHOOD/ROOM TYPE BY REVENUE PROXY (LOWER) ###")
#print(summary_lower.head(20).to_string(index=False))
print(summary_upper.head(20))
```

## Heatmap of Revenue Proxy by Neighborhood and Room Type

This exercise explores how `revenue_proxy_upper` varies across neighborhoods and room types. Using a heatmap provides an intuitive way to visualize patterns and trends across these two dimensions.

> 1. **Create a Heatmap**:
>    - Group the data by `neighbourhood_cleansed` and `room_type`.
>    - Filter combinations to include only those with at least 3 listings for meaningful averages.
>    - Compute the average `revenue_proxy_upper` for each valid combination.
> 
> 2. **Steps**:
>    - Pivot the grouped data into a matrix where rows represent neighborhoods and columns represent room types.
>    - Sort both rows (neighborhoods) and columns (room types) by their mean revenue values for better readability.
>    - Visualize the heatmap using the `viridis` color palette and annotate each cell with the average value.
> 
> 3. **Goal**:
>    - Identify neighborhoods and room types with the highest and lowest revenue proxies.
>    - Detect potential trends or anomalies in revenue generation across neighborhoods and room types.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def plot_revenue_heatmap(df):
    """
    Creates a prioritized heatmap of revenue across neighborhoods and room types.
    """
    # 1. Grouping and Filtering (Minimum 3 listings for statistical significance)
    grouped = df.groupby(['neighbourhood_cleansed', 'room_type']).agg(
        avg_revenue=('revenue_proxy_upper', 'mean'),
        listing_count=('name', 'count')
    ).reset_index()

    # Apply the threshold filter
    grouped_filtered = grouped[grouped['listing_count'] >= 3]

    # 2. Pivot the data into a matrix
    heatmap_data = grouped_filtered.pivot(
        index='neighbourhood_cleansed', 
        columns='room_type', 
        values='avg_revenue'
    )

    # 3. Sorting Rows and Columns by their mean values (The "Analyst" Sorting)
    # This ensures the high-revenue items cluster in the top-left
    row_order = heatmap_data.mean(axis=1).sort_values(ascending=False).index
    col_order = heatmap_data.mean(axis=0).sort_values(ascending=False).index
    
    heatmap_data = heatmap_data.loc[row_order, col_order]

    # 4. Visualization
    # Note: Berlin has 138 neighborhoods. To keep the map readable, 
    # we take the top 40 neighborhoods by average revenue.
    top_n = 20
    plot_matrix = heatmap_data.head(top_n)

    plt.figure(figsize=(14, 14)) # Tall figure for many neighborhoods
    sns.heatmap(
        plot_matrix, 
        annot=True,          # Show the revenue values
        fmt=".0f",           # No decimals for cleaner look
        cmap="viridis",      # Perceptually uniform palette
        cbar_kws={'label': 'Average Revenue Proxy (Upper) [€/Month]'},
        linewidths=.5
    )

    # 5. Customization
    plt.title(f"Top {top_n} Berlin Neighborhoods by Revenue Proxy (Upper)\n(Minimum 3 listings per segment)", 
              fontsize=16, pad=20)
    plt.xlabel("Room Type", fontsize=12)
    plt.ylabel("Neighborhood", fontsize=12)
    
    plt.tight_layout()
    plt.show()

# Run the heatmap function
plot_revenue_heatmap(df_final)
```

The heatmap shows that "Entire home/apt" generates the highest revenue across most neighborhoods, with standout areas marked by brighter hues. ## Missing license analysis

Verify the reason
1. The "Airbnb Enforcement" Hypothesis

Berlin has a very strict automated enforcement system with Airbnb. The platform usually prevents listings from being "Active" (bookable) if the license field is completely empty.

Because your filter required reviews_per_month > 2, you are looking at the most successful, active listings.
Hosts who want to stay active but don't have a license usually feel forced to type something in that box (e.g., "prive", "pending", "12345") rather than leaving it blank.
In our code, these "placeholder" strings are caught by the "Invalid/Dirty" logic, not the "Missing" logic.

2. The "Active Market" Filter Impact

It is highly probable that listings with a truly "Missing" license are either brand new or "shadowed" (hidden by Airbnb), meaning they have 0 reviews and were stripped away by our reviews_per_month > 2 filter.

```python
def prepare_revenue_data_v3(df):
    df_calc = df.copy()
    
    # 1. Price Cleaning
    df_calc['price_int'] = (
        df_calc['price'].replace(r'[\$,€%]', '', regex=True)
        .replace(',', '', regex=True).astype(float).fillna(0).astype(int)
    )

    # 2. Revenue Calculation
    DAYS_PER_MONTH = 30.44
    OCCUPANCY_CAP = 0.90
    df_calc['max_bookings'] = OCCUPANCY_CAP * (DAYS_PER_MONTH / df_calc['minimum_nights'])
    df_calc['revenue_proxy_upper'] = (df_calc['price_int'] * 
        df_calc[['reviews_per_month', 'max_bookings']].fillna(0).assign(
            est_bookings=df_calc['reviews_per_month']/0.3
        )[['est_bookings', 'max_bookings']].min(axis=1)).round(2)

    # 3. CRITICAL: Include 'license' in this list
    keep = [
        'name', 'room_type', 'neighbourhood_cleansed', 
        'minimum_nights', 'reviews_per_month', 'price_int',
        'revenue_proxy_upper', 'license' # <--- MUST BE HERE
    ]
    return df_calc[keep].rename(columns={'price_int': 'price'})

# Re-run the chain
revenue_df = prepare_revenue_data_v3(df) # df is your raw 79-column data

reviews_per_month_min = 2

df_filtered_2 = revenue_df[(revenue_df['reviews_per_month'] > 2) & (revenue_df['minimum_nights'] <= 5) & (revenue_df['price'] < 500)]


def check_missing_drift_safe(df_raw, df_filt, df_final):
    print("--- License Data Tracking ---")
    
    # Raw Check
    if 'license' in df_raw.columns:
        print(f"1. Raw Dataset:      {df_raw['license'].isna().sum()} listings have NaN license.")
    
    # Filtered Check
    if 'license' in df_filt.columns:
        filt_missing = df_filt['license'].isna().sum()
        print(f"2. Active Filter:    {filt_missing} listings have NaN license.")
    else:
        print("2. Active Filter:    ERROR - 'license' column was lost in this step!")

    # Final Check
    if 'license_status' in df_final.columns:
        final_missing = (df_final['license_status'] == 'Missing').sum()
        print(f"3. Final Table:      {final_missing} categorized as 'Missing'.")

# Execute
print("\nChecking with reviews_per_month > 2 filter:\n")
check_missing_drift_safe(df, df_filtered_2, df_final)

print("\nChecking with reviews_per_month > 0 filter:\n")
reviews_per_month_min = 0
df_filtered_0 = revenue_df[(revenue_df['reviews_per_month'] > 0) & (revenue_df['minimum_nights'] <= 5) & (revenue_df['price'] < 500)]
check_missing_drift_safe(df, df_filtered_0, df_final)
```

*   **Survival of the Fittest:** Truly "Missing" license data is almost non-existent in the high-performance market. This suggests that to "survive" the Airbnb algorithm in Berlin, hosts who lack a license feel forced to provide a "decoy" license number to maintain high booking frequency.

#### **C. Geographic Profitability Clusters**
The Heatmap analysis confirmed that **Mitte (Alexanderplatz, Brunnenstr. Süd)** and **Pankow (Prenzlauer Berg)** remain the revenue kings.
*   **Supply vs. Regulation:** Neighborhoods with high "Private Room" density and low "Entire Home" revenue indicate areas where city enforcement is likely more effective, forcing hosts into a lower-margin "Home-sharing" model.
